```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow import keras

print("Libraries Imported Successfully!\n")


(X_train_full, y_train_full), (X_test, y_test) = \
    keras.datasets.fashion_mnist.load_data()

print("Fashion MNIST Dataset Loaded Successfully!")
print("Training Images Shape:", X_train_full.shape)
print("Training Labels Shape:", y_train_full.shape)
print("Testing Images Shape:", X_test.shape)
print("Testing Labels Shape:", y_test.shape)


X_valid = X_train_full[:5000].astype("float32") / 255.0
X_train = X_train_full[5000:].astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

y_valid = y_train_full[:5000]
y_train = y_train_full[5000:]

print("Data Preprocessing Completed!")
print("Training Samples:", X_train.shape[0])
print("Validation Samples:", X_valid.shape[0])
print("Testing Samples:", X_test.shape[0])
print("Pixel Values Normalized Between 0 and 1\n")


class_names = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]

print("Classes:", class_names)
print("Total Number of Classes:", len(class_names))


model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(512, activation="relu"),
    keras.layers.Dense(300, activation="relu"),
    keras.layers.Dense(50, activation="relu"),
    keras.layers.Dense(10, activation="softmax")
])

print("MLP Model Created Successfully!")
print("Architecture: Flatten -> 512 -> 300 -> 50 -> 10")


model.summary()

print("\nModel Summary Displayed Successfully!\n")


model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

print("Model Compiled Successfully!")


history = model.fit(
    X_train,
    y_train,
    epochs=15,
    validation_data=(X_valid, y_valid),
    verbose=1
)

print("\nModel Training Completed Successfully!")


pd.DataFrame(history.history).plot(figsize=(8, 5))
plt.grid(True)
plt.gca().set_ylim(0, 1)
plt.title("Fashion MNIST Training and Validation Progress")
plt.xlabel("Epochs")
plt.ylabel("Value")
plt.show()

print("Training Graph Displayed Successfully!\n")


test_loss, test_acc = model.evaluate(
    X_test, y_test, verbose=1
)

print("\nModel Evaluation Completed!")
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc * 100:.2f}%")


X_new = X_test[:3]
y_true = y_test[:3]

y_proba = model.predict(X_new, verbose=0)

y_pred = np.argmax(y_proba, axis=1)

for i in range(len(X_new)):
    predicted = class_names[y_pred[i]]
    actual = class_names[y_true[i]]
    confidence = y_proba[i][y_pred[i]] * 100

    print(
        f"Image {i + 1}: Predicted [{predicted}], "
        f"Confidence: {confidence:.2f}%, "
        f"Actual: [{actual}]"
    )
```
